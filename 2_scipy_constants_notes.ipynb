## 1. What is SciPy?

**Short answer:** SciPy (Scientific Python) is a library of ready-made tools for maths, science, and statistics. It is built on top of NumPy.

**Real-life example:** NumPy is like a good calculator. SciPy is like a toolbox of special machines that use that calculator inside. You do not build the machine yourself. You just use it.

**How it connects to what you already know:**

| Library | What it gives you |
|---|---|
| NumPy | fast arrays and basic maths |
| Pandas | tables (DataFrames) |
| SciPy | scientific and statistics tools that work on NumPy arrays and Pandas columns |

### SciPy modules in this W3Schools tutorial

| Topic | Job (simple words) | How useful for me |
|---|---|---|
| Constants | ready-made numbers and unit values | low (but easy) |
| Optimizers | find the best (minimum) value of something | medium. This is how ML models learn |
| Sparse Data | store data that is mostly zeros | medium |
| Graphs | paths and connections in networks | low |
| Spatial Data | points, distance, nearest neighbours | medium (clustering later) |
| Matlab Arrays | read and write MATLAB files | low |
| Interpolation | guess values between known points | medium (missing data) |
| Significance Tests (`scipy.stats`) | t-test, p-value, distributions | **very high** |

**Important:** `scipy.stats` is the part that matters most for data analysis and ML interviews. Constants is only the warm-up.

## 3. What is `scipy.constants`?

**Short answer:** It is a box of **correct, ready-made numbers** for science and unit conversion, so you do not type them by hand.

**Real-life example:** In a shipping office there is a conversion chart on the wall: 1 mile = 1609 meters, 1 pound = 0.45 kg. Nobody remembers all numbers. They just look at the chart. `scipy.constants` is that chart, inside Python.

**Why use it?**
1. **No typing mistakes.** If you write `3600` in 10 places and one is `3060`, your answer is wrong and you may never see it.
2. **Clear code.** `days * constants.day` is easier to read than `days * 86400`.
3. **Exact values.** You get the full number, not a rounded one like `1.6`.

### First example: pi

````python
from scipy import constants     # load the constants module
import numpy as np

print(constants.pi)             # 3.141592653589793
print(np.pi)                    # same value from NumPy
````

### See everything inside the module: `dir()`

`dir(constants)` lists the names inside the module. Use it when you forget a name. The names starting with `_` are hidden helper names, so we skip them.

````python
names = [n for n in dir(constants) if not n.startswith('_')]   # keep only normal names
print(len(names))      # how many names are inside (changes with the SciPy version)
print(names[:15])      # show the first 15 names
````

## 4. Base units and the maths of conversion

**The big idea:** Every unit constant is stored in a **base unit**. The constant tells you: *how many base units are in ONE of this unit?*

| Category | Base unit | Example |
|---|---|---|
| Time | seconds | `constants.hour` = 3600.0 |
| Length | meters | `constants.mile` = 1609.344 |
| Mass | kilograms | `constants.pound` = 0.4536 |
| Area | square meters | `constants.hectare` = 10000.0 |
| Volume | cubic meters | `constants.liter` = 0.001 |
| Speed | meters per second | `constants.kmh` = 0.2778 |
| Pressure | pascals | `constants.bar` = 100000.0 |
| Temperature | Kelvin | `constants.zero_Celsius` = 273.15 |
| Energy | joules | `constants.calorie` = 4.184 |
| Power | watts | `constants.hp` = 745.7 |
| Force | newtons | `constants.kgf` = 9.80665 |
| Angle | radians | `constants.degree` = 0.01745 |

### The maths (simple)

1. Go **into** the base unit: `value_in_base = value * constants.unit`
2. Come **out** of the base unit: `value_in_unit = value_in_base / constants.unit`
3. From unit A to unit B in one line: `value * constants.A / constants.B`

**Memory trick:** *Multiply to go in, divide to come out.*

**Example by hand:** 5 miles to km.
- 5 miles x 1609.344 = 8046.72 meters (go in)
- 8046.72 / 1000 = 8.04672 km (come out, because `kilo` = 1000)

````python
from scipy import constants

miles = 5
km = miles * constants.mile / constants.kilo   # miles -> meters -> km
print(km)                                      # 8.046719999999999 (tiny float noise is normal)
print(round(km, 2))                            # 8.05
````

## 5. Metric prefixes (SI prefixes)

These are **plain numbers** that you multiply or divide by.

| Name | Value | Meaning |
|---|---|---|
| `giga` | 1e9 | one billion |
| `mega` | 1e6 | one million |
| `kilo` | 1000 | one thousand |
| `milli` | 0.001 | one thousandth |
| `micro` | 1e-06 | one millionth |
| `nano` | 1e-09 | one billionth |

**Note:** The W3Schools page says these return a unit 'in meter'. That wording is confusing. The prefixes are just pure numbers, for any unit.

**Real-life example:** A sales report has revenue of 125,000,000 PKR. The manager wants it in **millions**. A web page loads in 0.25 seconds. The developer wants it in **milliseconds**.

````python
from scipy import constants

revenue_pkr = 125_000_000
print(revenue_pkr / constants.mega)    # 125.0  -> 125 million PKR

load_time_s = 0.25
print(load_time_s / constants.milli)   # 250.0  -> 250 milliseconds
````

## 6. Binary prefixes

Computers count in powers of **2**, not 10. So a 'kilo' of memory is 1024, not 1000. These prefixes give values in **bytes**.

| Name | Value |
|---|---|
| `kibi` | 1024 |
| `mebi` | 1,048,576 (1024 x 1024) |
| `gibi` | 1,073,741,824 |

**Interview point:** `kilo` = 1000 and `kibi` = 1024. They are close but not the same.

**Real-life example:** A client sends a data file. You want to know its size in MB.

````python
from scipy import constants

file_bytes = 5_242_880
print(file_bytes / constants.mebi)     # 5.0  -> 5 MB

print(2 * constants.gibi)              # 2147483648 -> bytes in 2 GB (a whole number)
````

## 7. Real-life example 1: Hotel Booking (time)

In your Hotel Booking project, `lead_time` is in **days**. Suppose a manager asks for lead time in **hours** and **weeks**.

Time constants (all in seconds): `minute` = 60, `hour` = 3600, `day` = 86400, `week` = 604800, `year` = 31536000 (365 days).

**Plan:** days -> seconds (multiply by `day`), then seconds -> hours or weeks (divide). This is Pandas column maths, which you already know.

````python
import pandas as pd
from scipy import constants

df = pd.DataFrame({'lead_time_days': [3, 30, 120, 365]})   # small fake lead_time column

df['seconds'] = df['lead_time_days'] * constants.day       # days -> seconds (go in: multiply)
df['hours'] = df['seconds'] / constants.hour               # seconds -> hours (come out: divide)
df['weeks'] = df['seconds'] / constants.week               # seconds -> weeks

print(df.round(2))
````

## 8. Real-life example 2: Freelance client file (length and mass)

A US client sends delivery data with distance in **miles** and weight in **pounds**. Your report must be in **km** and **kg**.

- `constants.mile` is in meters, so divide by `constants.kilo` to get km.
- `constants.pound` is already in kg, so just multiply.

````python
import pandas as pd
from scipy import constants

orders = pd.DataFrame({
    'order_id': [101, 102, 103],
    'distance_miles': [10, 25, 100],
    'weight_pounds': [5, 20, 150]
})

orders['distance_km'] = orders['distance_miles'] * constants.mile / constants.kilo   # miles -> km
orders['weight_kg'] = orders['weight_pounds'] * constants.pound                      # pounds -> kg

print(orders.round(2))
````

## 9. Speed and temperature

**Speed:** `constants.kmh` and `constants.mph` are both in meters per second. So km/h to mph is `value * kmh / mph`.

**Temperature is special.** Fahrenheit and Celsius do not start at the same zero. So multiplying alone is **not enough**. You also need an **offset** (the starting point):

- `constants.degree_Fahrenheit` = 0.5556 (the size of one Fahrenheit degree, in Kelvin)
- `constants.zero_Celsius` = 273.15 (0 Celsius, in Kelvin)

Formula: `Kelvin = (F - 32) * degree_Fahrenheit + zero_Celsius`, then `Celsius = Kelvin - zero_Celsius`.

**Real-life example:** A car speed of 100 km/h in miles per hour, and a body temperature of 98.6 F in Celsius.

````python
from scipy import constants

# speed: 100 km/h -> mph
speed_mph = 100 * constants.kmh / constants.mph
print(round(speed_mph, 2))    # 62.14

# temperature: 98.6 F -> C
f = 98.6
kelvin = (f - 32) * constants.degree_Fahrenheit + constants.zero_Celsius
celsius = kelvin - constants.zero_Celsius
print(round(celsius, 2))      # 37.0
````

## 10. Other categories (quick examples)

| Category | Real-life example |
|---|---|
| Volume | 10 gallons of petrol in liters |
| Pressure | car tyre at 32 psi in bar |
| Energy | 500 food Calories in joules (a food Calorie is 1 kcal = 1000 calories) |
| Angle | `np.sin()` needs **radians**, so change degrees first |

````python
import numpy as np
from scipy import constants

# volume: 10 gallons -> liters
print(round(10 * constants.gallon / constants.liter, 2))   # 37.85

# pressure: tyre 32 psi -> bar
print(round(32 * constants.psi / constants.bar, 2))        # 2.21

# energy: 500 food Calories (kcal) -> joules
print(500 * constants.kilo * constants.calorie)            # 2092000.0

# angle: sin(90 degrees) must use radians
print(np.sin(90 * constants.degree))                       # 1.0
````

## 12. Common mistakes

1. **Thinking `constants.hour` is 1.** It is 3600, because it is stored in seconds.
2. **Multiplying when you should divide.** Remember: multiply to go in, divide to come out.
3. **Mixing `kilo` (1000) and `kibi` (1024).** Use `kibi`, `mebi`, `gibi` for computer memory.
4. **Converting temperature by multiplying only.** You need the offset too.
5. **Using degrees in `np.sin()`.** Multiply by `constants.degree` first.
6. **Mixing `pound` and `pound_force`.** `pound` is mass (kg). `pound_force` is force (newton).

## 14. Interview questions (try to answer first, then read)

**Q1. What is SciPy and how is it different from NumPy?**
SciPy is built on NumPy. NumPy gives arrays and basic maths. SciPy adds scientific tools such as statistics, optimization, and interpolation.

**Q2. What is `scipy.constants` used for?**
It gives exact, ready-made values for units and science constants, so we can convert units without typing numbers by hand.

**Q3. What is the base unit for time, length, and mass?**
Seconds, meters, and kilograms.

**Q4. What is the difference between `kilo` and `kibi`?**
`kilo` is 1000 (power of 10). `kibi` is 1024 (power of 2, used for computer memory).

**Q5. How do you convert 5 miles to km with SciPy?**
`5 * constants.mile / constants.kilo`, which is about 8.05 km.

**Q6. Why can you not change Fahrenheit to Celsius by only multiplying?**
Because the two scales have different zero points. You need to subtract 32 (the offset) as well as scale the size of the degree.

**Q7. How do you see all names in the constants module?**
`print(dir(constants))`.
